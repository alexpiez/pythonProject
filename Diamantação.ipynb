```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

data = {
  'altitude': [35000, 38000, 32000, 40000],
  'umidade': [20, 10, 50, 30],
  'particulas': [0.1, 0.5, 0.01, 0.3],
  'diamantacao': [0, 1, 0, 1]
}
df = pd.DataFrame(data)

X = df[['altitude', 'umidade', 'particulas']]
y = df['diamantacao']
rf = RandomForestClassifier()
rf.fit(X, y)

new_flight = pd.DataFrame([[37000, 15, 0.2]], columns=X.columns) # 👈 add columns
pred = rf.predict_proba(new_flight)
print(f"Probabilidade de diamantação: {pred[0][1]:.2f}")
```

```python
# Dados fake
cidades = {
  'Miami': {'umidade': 80, 'poluição': 20, 'altitude': 0},
  'Toronto': {'umidade': 50, 'poluição': 70, 'altitude': 100}
}
rota = {'altitude_media': 8000, 'poluição_rota': 50}

# Modelo tosco
def diamantacao(origem, destino, rota):
  score = (origem['poluição'] + destino['poluição'] + rota['poluição_rota']) / 300
  score += rota['altitude_media'] / 10000
  return score

miami = cidades['Miami']
toronto = cidades['Toronto']
prob = diamantacao(miami, toronto, rota)
print(f"Diamantação Miami->Toronto: {prob:.2f}")
```
